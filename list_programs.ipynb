```python
#9
nums = []
for i in range(20):
    num = int(input(f"enter number {i+1}: "))
    nums.append(num) 

# part a
numslist = []
for i in range(20):
    current_num = nums[i]
    if nums.count(current_num) > 1 and current_num not in numslist:
        numslist.append(current_num)
        indices = []
        for j in range(20):
            if nums[j] == current_num:
                indices.append(j)
        print(f"Number {current_num} appears {nums.count(current_num)} times at indices: {indices}")

# part b
odd = []
even = []
for i in range(20):
    current_num = nums[i]
    if current_num % 2 == 0:
        even.append(current_num)
    else:
        odd.append(current_num)
        
print(f"List of even numbers is: {even}")
print(f"List of odd numbers is: {odd}")      

# part c
positive = []
negative = []
zero = []
for i in range(20):
    current_num = nums[i]
    if current_num > 0:
        positive.append(current_num)
    elif current_num < 0:
        negative.append(current_num)
    else: 
        zero.append(current_num)

print(f"List of positive numbers is: {positive}")
print(f"List of negative numbers is: {negative}")  
print(f"List of zeroes is: {zero}")
```
