```python
#creating a series from a list
import pandas as pd
a = [1,7,2]
myvar = pd.Series(a)
print(myvar)
```

```python
#creating labels
import pandas as pd
a = [1,7,2]
myvar = pd.Series(a , index = ["x" , "y" , "z"])
print(myvar)
```

```python
#creating a series
import pandas as pd
a = [10,20,30]
myvar = pd.Series(a , index = ["a" , "b" , "c"] , name = "example series")
print(myvar)
```

```python
#creating a series from a dictionary
import pandas as pd
s1 = pd.Series({'x':1 , 'y':2 , 'z':3})
print(s1)
print()
calories = {'day1':240 , 'day2':380 , 'day3':390}
myvar = pd.Series(calories)
print(myvar)
```

```python
#dataframes
import pandas as pd
data = {
    "calories":[420,380,390],
    "duration":[50,40,45]
}
myvar = pd.DataFrame(data)
print(myvar)
```

```python
# adding labels to dataframes 
import pandas as pd
data = {
    "calories":[420,380,390],
    "duration":[50,40,45]
}
df = pd.DataFrame(data)

#locating row
print(df.loc[0])
print()

#return row 0 and 1
print(df.loc[[0, 1]])
```

```python
#named indexes
import pandas as pd
data = {
    "calories":[420,380,390],
    "duration":[50,40,45]
}
df = pd.DataFrame(data , index = ["day1","day2","day3"])
print(df)
```

```python
#cleaning data (removing rows with missing values)
import pandas as pd
data = {
    'A':[1,2,3,None,5],
    'B':[1,2,None,None,5],
    'C':[None,2,3,4,5]
}
df = pd.DataFrame(data)
print("original data:\n",df)
print()
df_cleaned = df.dropna()
print("cleaned data:\n",df_cleaned)
```

```python
# filling missing values
import pandas as pd
data = {
    'A':[1,2,3,None,5],
    'B':[1,2,None,None,5],
    'C':[None,2,3,4,5]
}
df = pd.DataFrame(data)
print("original data:\n",df)
print()
df_cleaned = df.fillna(0, inplace=True)
print("cleaned data:\n",df_cleaned)
```

```python
# filling using aggregate functions
import pandas as pd
data = {
    'A':[1,2,3,None,5],
    'B':[1,2,None,None,5],
    'C':[None,2,3,4,5]
}
df = pd.DataFrame(data)
print("original data:\n",df)
print()
df_cleaned = df.fillna(df.mean(), inplace=True)
print("filled data:\n",df_cleaned)
```
