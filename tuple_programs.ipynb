```python
#4
nums = (10, 20, 30, 40, 50)
element = int(input("enter element:"))
if element in nums: 
    print("index of element:", nums.index(element))
else:
    print("not found")
```

```python
#5
nums = [(10, 20, 40), (70, 60, 20), (30, 50, 70)]
new_list=[]
for tup in nums:
    new_tup = tup[:-1] + (100,)
    new_list.append(new_tup)
print("original list:", nums)
print("updated list:", new_list)
```

```python
#6
nums = (12, 55, 78, 2, 45, 90, 0)
minimum = min(nums)
maximum = max(nums)
print("minimum element: ", minimum)
print("maximum element: ", maximum)
```

```python
#7
nums = (10, 20, 30, 40, 50)  
new_tuple = (nums[-1],) + nums[1:-1] + (nums[0],)  
print("original Tuple:", nums)  
print("tuple after swapping:", new_tuple)  
```
