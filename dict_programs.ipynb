```python
#3
mydict = {1:"rahul" , 2:"shreya" , 3:"anand"}
key=int(input("enter key to search:"))
if key in mydict:
    print("key exists in dictionary")
else:
    print("key does not exist in dictionary")
```

```python
#4
mydict = {1:"rahul" , 2:"shreya" , 3:"anand"}
print("KEYS:")
for key in mydict:      #mydict.keys() can also be used but not needed
    print(key)
print("VALUES:")
for value in mydict.values():
    print(value)
print("KEYS AND VALUES:")
for key,value in mydict.items():
    print(key, value)
```

```python
#5
mydict = {}
for i in range(1, 16):
    mydict[i] = i**2
print(mydict)
```

```python
#6
mydict = {"a":10 , "b":20 , "c":30 , "d":40}
total = 0
for value in mydict.values():
    total = total +value
print("sum of values:", total)
```
